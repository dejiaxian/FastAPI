### Yolo Model Training

```python
from ultralytics import YOLO
import torch

# Initialize YOLO model
model = YOLO("yolov8n.pt")  # Use YOLOv8 Nano (small and fast)

# Check if CUDA is available, else use CPU
device = 'cuda' if torch.cuda.is_available() else 'cpu'

# Train the model
model.train(
    data="IKad_Temp_Cropping.yaml",    
    epochs=15,                          # Number of epochs
    imgsz=640,                          # Image size
    batch=16,                           # Batch size
    name="IKad_Temp_Cropping_model",    # Training run name
    device=device,                      # Use GPU or CPU
    val=True                            # Disable validation during training
)

# Test the model using the test dataset after training
model.test(data="IKad_Temp_Cropping.yaml", device=device)
 
```
